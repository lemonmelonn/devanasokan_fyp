```python
# Import libraries
import pandas as pd
import re
```

```python
# Load dataset
df = pd.read_csv("../preparation/cleanverses.csv", encoding='utf-8-sig')
print("Initial shape:", df.shape)
```

```python
# Unique values in 'label' column
labels = df['label'].unique()
print("Unique labels:", labels.tolist())
print("Total unique labels:", len(labels))
```

```python
# Only keep rows where 'label' is SAFE or UNSAFE
df = df[df['label'].isin(['SAFE', 'UNSAFE'])]
print("Unique labels after filtering:", df['label'].unique().tolist())
print("Shape after filtering labels:", df.shape)
```

```python
# Map 'explicit' column from 0/1 to No/Yes
df['explicit'] = df['explicit'].map({0: "No", 1: "Yes"})
print("Unique labels after filtering:", df['explicit'].unique().tolist())
```

```python
# Clean 'section' column by keeping section names
def clean_section(section):
    if pd.isna(section):
        return section
    match = re.search(r'\b(Verse|Chorus|Bridge|Intro|Outro|Pre-Chorus|Hook|Interlude)\b', section, re.IGNORECASE)
    return match.group(0).capitalize() if match else "Other"

df['section'] = df['section'].apply(clean_section)
print("Unique sections after cleaning:", df['section'].unique().tolist())
```

```python
# Impute missing values in 'main_genre'
df['main_genres'] = df['main_genres'].fillna("Unknown")
print(df['main_genres'].isnull().sum())
```

```python
# Save output
print("After cleaning:", df.shape)
df.to_csv('../preparation/cleanverses.csv', index=False, encoding='utf-8-sig')
```

```python
# Check if column has normal brackets
def has_normal_brackets(text):
    return bool(re.search(r"\(.*?\)", text))

for index, row in df.iterrows():
    if has_normal_brackets(row['lyrics']):
        print(f"Row {index} has normal brackets: {row['lyrics']}")
```
