```python
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("C:/Users/User/Documents/devanasokan_fyp/preparation/cleanverses.csv", encoding='utf-8-sig')
print("Initial shape:", df.shape)
```

```python
print(df.columns.tolist())
```

```python
# Mapping labels to integers
label_mapping = {"SAFE": 0, "UNSAFE": 1}
df["label"] = df["label"].map(label_mapping)
```

```python
# Rename columns for clarity
df.rename(columns={"track_name": "ori_track_name", 
                   "artist_names": "all_artists",
                   "main_genres": "main_genre",
                   "verse_language": "language",
                   "verse_confidence": "confidence",
                   "lyrics": "verse"}, inplace=True)
```

```python
# Reduce decimal places for confidence
df['confidence'] = df['confidence'].round(2)
```

```python
# Reorder columns for better readability
df = df[['verse_id', 'song_id', 'ori_track_name', 'clean_track_name', 
         'all_artists', 'primary_artist', 'artist_genres', 'main_genre', 
         'explicit', 'section', 'verse', 'language', 'confidence', 'label']]
```

```python
# Save output
print("After tokenizing:", df.shape)
df.to_csv('C:/Users/User/Documents/devanasokan_fyp/preparation/finalverses.csv', index=False, encoding='utf-8-sig')
```
