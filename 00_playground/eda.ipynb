```python
import pandas as pd
import ast
```

```python
df = pd.read_csv("C:/Users/User/Documents/devanasokan_fyp/datasource/top-10k-spotify-songs-2025-07-detailed.csv")
```

```python
# Display track names and artists
print(df[["track_name", "artist_names"]].head(5))
```

```python
print(df.columns)
print("Number of columns:", len(df.columns))
```

```python
# Only keep relevant columns
df = df[["rank", "track_name", "artist_names", "album_name", 
         "explicit", "artist_genres", "main_genres", "lyrics"]]
```

### Duplicates

```python
# Check for duplicates
print(df.duplicated().sum())
```

```python
# Identify rows where both 'track_name' and 'artist_names' are identical
duplicate_mask = df.duplicated(subset=['track_name', 'artist_names'], keep='first')

# View the duplicate rows
duplicate_rows = df[duplicate_mask]
print(f"Number of duplicate songs: {len(duplicate_rows)}")
```

```python
# Returns a summary of duplicate counts for each column independently
duplicate_summary = {col: df[col].duplicated().sum() for col in df.columns}
print(pd.Series(duplicate_summary, name="Duplicate Count"))
```

```python
# Returns the count of unique values for each column
unique_summary = df.nunique()
print(unique_summary)
```

```python
duplicate_lyrics = df[df.duplicated(subset=['lyrics'], keep=False)]
print(duplicate_lyrics[["track_name", "artist_names", "lyrics"]])
```

### Missing Values

```python
missing_counts = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Count': missing_counts,
    'Percentage (%)': missing_percentage
})

# Display the summary sorted by missing values (descending)
print(missing_summary.sort_values(by='Missing Count', ascending=False))
```

```python
# Convert the 'artist_genres' column from string representation of list to actual list
df['genre'] = df['artist_genres'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)
genre_counts_ori = df['genre'].explode().value_counts()
print(genre_counts_ori)
```

```python
# Create a DataFrame containing only rows where 'lyrics' is missing
df_missing_lyrics = df[df['lyrics'].isna()].copy()

print(f"Number of songs with missing lyrics: {len(df_missing_lyrics)}")

# Convert the 'artist_genres' column from string representation of list to actual list
df_missing_lyrics['genre'] = df_missing_lyrics['artist_genres'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)
unique_genres = df_missing_lyrics['genre'].explode().unique()
unique_genres_list = sorted([str(g) for g in unique_genres if pd.notna(g)])

print(unique_genres_list)
print(len(unique_genres_list))

genre_counts_missing = df_missing_lyrics['genre'].explode().value_counts()

print(genre_counts_missing)
```

```python
# Determine which genres and counts match between the original and missing lyrics datasets
matching_genres = pd.merge(genre_counts_ori, genre_counts_missing, on=['genre', 'count'], how='inner')

print("List of genres with matching counts:")
genre_list = matching_genres.index.tolist()
print(genre_list)
print("Number of matching genres: " + str(len(genre_list)))
```

```python
missing_counts = df_missing_lyrics.isnull().sum()
missing_percentage = (df_missing_lyrics.isnull().sum() / len(df_missing_lyrics)) * 100

missing_summary = pd.DataFrame({
    'Missing Count': missing_counts,
    'Percentage (%)': missing_percentage
})

# Display the summary sorted by missing values (descending)
print(missing_summary.sort_values(by='Missing Count', ascending=False))
```
