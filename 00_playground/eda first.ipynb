### Duplicates

```python
# Check for duplicates
print(df.duplicated().sum())
```

```python
# Returns a summary of duplicate counts for each column independently
duplicate_summary = {col: df[col].duplicated().sum() for col in df.columns}
print(pd.Series(duplicate_summary, name="Duplicate Count"))
```

```python
# Returns the count of unique values for each column
unique_summary = df.nunique()
print(unique_summary)
```

```python
# Identify rows where both 'track_name' and 'artist_names' are identical
duplicate_mask = df.duplicated(subset=['track_name', 'artist_names'], keep='first')

# View the duplicate rows
duplicate_rows = df[duplicate_mask]
print(f"Number of duplicate songs: {len(duplicate_rows)}")
print(duplicate_rows.head())
```

### Missing Values

```python
missing_counts = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Count': missing_counts,
    'Percentage (%)': missing_percentage
})

# Display the summary sorted by missing values (descending)
print(missing_summary.sort_values(by='Missing Count', ascending=False))
```
