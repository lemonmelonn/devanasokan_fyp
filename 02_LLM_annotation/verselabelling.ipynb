```python
import pandas as pd
from tqdm import tqdm
import time
import json
import ollama
```

```python
df = pd.read_csv("C:/Users/User/Documents/devanasokan_fyp/storage/splitverses.csv", encoding='utf-8-sig')
```

```python
print(df.shape)
print(df.columns)
```

```python
# Prompt template
def create_prompt(verse):

    prompt = f"""
    You are an AI assistant helping with a dataset annotation task for a research project.

    TASK: Your task is to classify song verses based on whether they are appropriate for children to listen to.

    Label the verse as:

    UNSAFE
    * Sexual or suggestive content
    * Explicit language or profanity
    * References to drugs or alcohol

    SAFE
    * No mention of any topics listed above

    Only respond with one word: SAFE or UNSAFE.


    Verse:
    "{verse}"
    """

    return prompt

```

```python
labels = []

# Loop through dataset
for verse in tqdm(df["genius_lyrics"]):

    try:

        response = ollama.chat(
            model="llama3",   # change if needed
            messages=[
                {"role": "user", "content": create_prompt(verse)}
            ]
        )

        label = response["message"]["content"].strip()

    except Exception as e:
        label = "ERROR"

    # Store label in list
    labels.append(label)

# Add labels to dataframe
df["label"] = labels
```

```python
print(labels)
print(len(labels))
```

```python
# Add labels to dataframe
df["label"] = labels

# Save result
df.to_csv("C:/Users/User/Documents/devanasokan_fyp/storage/labeledverses.csv", index=False)
```
