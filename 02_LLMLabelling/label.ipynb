```python
df = pd.read_csv("C:/Users/User/Documents/devanasokan_fyp/storage/englishsongs.csv", encoding='utf-8-sig')
```

```python
# Prompt template
def create_prompt(lyric):

    prompt = f"""
You are an AI assistant helping with a dataset annotation task for a research project.

TASK: Your task is to classify song lyrics based on whether they are appropriate for children to listen to.

Label the lyrics as:

UNSAFE
* Sexual or suggestive content
* Explicit language or profanity
* References to drugs or alcohol

SAFE
* No mention of any topics listed above

Only respond with one word: SAFE or UNSAFE.




Verse:
"{lyric}"
"""

    return prompt

# try return extra value (reason, confidence score?)
```

```python
labels = []

# Loop through dataset
for lyric in tqdm(df["lyrics"]):

    try:

        response = ollama.chat(
            model="llama3",   # change if needed
            messages=[
                {"role": "user", "content": create_prompt(lyric)}
            ]
        )

        label = response["message"]["content"].strip()

    except Exception as e:
        label = "ERROR"

    labels.append(label)
```

```python
print(labels)
```

```python
# Add labels to dataframe
df["label"] = labels
```

```python
# Save result
df.to_csv("labeled_songs.csv", index=False)
```
