```python
from functions import load_model ,detect_explicit, get_structured_lyrics, clean_verses ,get_model_output
from spotify_functions import get_access_token, get_song_details
from csv_functions import add_nonexplicit_song, add_explicit_song, check_song_exists, update_song_label, update_song_info
```

```python
from pathlib import Path
print(Path("05_model_training/my_final_model").resolve())
print(Path("05_model_training/my_final_model").exists())
```

```python
# Load the model and tokenizer from your local folder
from transformers import AutoModelForSequenceClassification, AutoTokenizer, pipeline


path = "./05_model_training/my_final_model"
model = AutoModelForSequenceClassification.from_pretrained(path)
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased") # Ensure tokenizer has been saved

# Create a 'pipeline' (the easiest way to use the model)
classifier = pipeline("sentiment-analysis", model=model, tokenizer=tokenizer)
```

```python
token = get_access_token()
```

```python
song = get_song_details(
    song_title="Praise God",
    artist_name="Kanye West",
    access_token=token
)
```

```python
print(song)
SONG_ID = song["song_id"]
ARTIST = song["artist"]
TITLE = song["title"]
```

```python
explicit = detect_explicit(song)
if explicit:
    CSV_FILE = "explicit.csv"
else:
    CSV_FILE = "nonexplicit.csv"

exists = check_song_exists(SONG_ID, CSV_FILE)
```

```python
def non_explicit_pipeline(title, artist):
    full_song = get_structured_lyrics(artist, title)
    print(full_song[:100])

    verses = clean_verses(full_song)
    for verse in verses["lyrics"]:
        get_model_output(classifier, [verse])
        print("\n---\n")
```

```python
if not exists:
    if explicit:
        add_explicit_song(SONG_ID, TITLE, ARTIST, CSV_FILE)
    else:
        add_nonexplicit_song(SONG_ID, TITLE, ARTIST, CSV_FILE)
        non_explicit_pipeline(TITLE, ARTIST)
else:
    print(f"Song ID {SONG_ID} already exists in {CSV_FILE}.")
```

```python
update_song_label(SONG_ID, "UNSAFE", CSV_FILE)
```
